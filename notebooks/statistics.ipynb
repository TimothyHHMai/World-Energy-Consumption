# Plots

```python
import plotly.express as px

def rmse_plot(dataset):
    
    fig_rmse = px.bar(
    dataset,
    x='Model',
    y='RMSE',
    color='Dataset',
    barmode='group',
    text_auto='.2f',
    title='RMSE Comparison'
    )
    
    fig_rmse.show()
    
def smape_plot(dataset):
    fig_smape = px.bar(
    dataset,
    x='Model',
    y='sMAPE',
    color='Dataset',
    barmode='group',
    text_auto='.2f',
    title='sMAPE Comparison'
    )

    fig_smape.show()
```
