## Libraries

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## Data Acquisition

```python
path = "../data/World Energy Consumption.csv"

df = pd.read_csv(path)

df.head()
```

## EDA

```python
corr = df.corr(numeric_only=True)["primary_energy_consumption"].sort_values(ascending=False)
print(corr)
```

```python
renewable_cols = [
    'solar_consumption',
    'wind_consumption',
    'hydro_consumption',
    'renewables_consumption'
]

plt.figure(figsize=(6,4))
plt.title("Correlation Heatmap - Renewable")

sns.heatmap(df[renewable_cols].corr(), annot=True, cmap='coolwarm')

plt.show()
```

```python
fossil_cols = [
    'coal_consumption',
    'oil_consumption',
    'gas_consumption',
    'fossil_fuel_consumption',
    'co2_emissions'
]

plt.figure(figsize=(6,4))
plt.title("Correlation Heatmap - Fossil")

sns.heatmap(df[renewable_cols].corr(), annot=True, cmap='coolwarm')

plt.show()
```

```python
global_energy = df.groupby('year')['primary_energy_consumption'].sum()

plt.figure(figsize=(10,6))
plt.plot(global_energy.index, global_energy.values)

plt.title("Global Energy Consumption Over Time")
plt.xlabel("Year")
plt.ylabel("Energy Consumption")
plt.show()
```

```python
df_num = df.select_dtypes(include="number")

for col in df_num.columns:
    
    plt.figure(figsize=(8,5))
    
    plt.hist(df_num[col].dropna(), bins=50)
    
    plt.title(f"Distribution of {col}")
    
    plt.show()
```
