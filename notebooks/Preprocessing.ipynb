## Libraries

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

## Data Acquisition
https://www.kaggle.com/datasets/pralabhpoudel/world-energy-consumption/data

```python
path = "../data/World Energy Consumption.csv"

df = pd.read_csv(path)

df.head()
```

## Data Cleaning

### Values with '?'

```python
cols_with_question = df.columns[df.isin(['?']).any()].tolist()
print(cols_with_question)
```

### Missing Values

```python
missing_pct = (df.isna().sum() / len(df)) * 100

print(missing_pct[missing_pct > 80])
```

Drop columns over 80% missing values.

```python
cols_to_drop = missing_pct[missing_pct > 80].index

df = df.drop(columns=cols_to_drop)
```

## EDA

```python
corr = df.corr(numeric_only=True)["primary_energy_consumption"].sort_values(ascending=False)
print(corr)
```

```python
renewable_cols = [
    'solar_consumption',
    'wind_consumption',
    'hydro_consumption',
    'renewables_consumption'
]

plt.figure(figsize=(6,4))
plt.title("Correlation Heatmap - Renewable")

sns.heatmap(df[renewable_cols].corr(), annot=True, cmap='coolwarm')

plt.show()
```

```python
fossil_cols = [
    'coal_consumption',
    'oil_consumption',
    'gas_consumption',
    'fossil_fuel_consumption',
    'co2_emissions'
]

plt.figure(figsize=(6,4))
plt.title("Correlation Heatmap - Fossil")

sns.heatmap(df[renewable_cols].corr(), annot=True, cmap='coolwarm')

plt.show()
```

```python
global_energy = df.groupby('year')['primary_energy_consumption'].sum()

plt.figure(figsize=(10,6))
plt.plot(global_energy.index, global_energy.values)

plt.title("Global Energy Consumption Over Time")
plt.xlabel("Year")
plt.ylabel("Energy Consumption")
plt.show()
```

```python
df_num = df.select_dtypes(include="number")

for col in df_num.columns:
    
    plt.figure(figsize=(8,5))
    
    plt.hist(df_num[col].dropna(), bins=50)
    
    plt.title(f"Distribution of {col}")
    
    plt.show()
```

## Feature Selection

```python
features = [
    'primary_energy_consumption', # Target
    'population',
    'gdp',
    'coal_consumption',
    'oil_consumption',
    'gas_consumption',
    'fossil_fuel_consumption',
    'solar_consumption',
    'wind_consumption',
    'hydro_consumption',
    'renewables_consumption',
    'electricity_generation',
    'hydro_share_energy'
]
```

```python
df = df[features]
```
